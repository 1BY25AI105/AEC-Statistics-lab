```python
# LAB PROGRAM 2 (Question 1-4)
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

n, p = 20, 0.05
binom = stats.binom(n, p)
x = np.arange(0, n+1)

print(f' Mean = {binom.mean():.4f} (Theory: np={n*p})')
binom.var()
binom.moment(2)
binom.stats("s")
binom.cdf(10)
binom.pmf(7)

#Visualisation

plt.figure(figsize=(7, 5))
plt.bar(x, binom.pmf(x), color='steelblue',
        edgecolor='black', alpha=0.8)

plt.axvline(binom.mean(), color='red', linestyle='--',
            label=f'Mean = {binom.mean():.1f}')

plt.title(f'Binomial PMF (n={n}, p={p})')
plt.xlabel('Number of Successes (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# LAB PROGRAM 2 (Question 6)
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

lam = 5
n = 100
p = lam / n

x = np.arange(0, 15)

poisson_probs = stats.poisson.pmf(x, lam)
binomial_probs = stats.binom.pmf(x, n, p)
print(f"{'x':<5} | {'Poisson Prob':<15} | {'Binomial Prob':<15}")
print("-" * 41)
for xi, p_pois, p_binom in zip(x, poisson_probs, binomial_probs):
    print(f"{xi:<5} | {p_pois:<15.5f} | {p_binom:<15.5f}")

# Visualization
plt.figure(figsize=(8, 5))
plt.plot(x, poisson_probs, 'o-', label=f'Poisson ($\lambda$={lam})', color='blue')
plt.plot(x, binomial_probs, 's--', label=f'Binomial ($n$={n}, $p$={p})', color='orange')
plt.title('Comparison: Poisson Distribution vs. Binomial Approximation')
plt.xlabel('Number of Events / Successes ($x$)')
plt.ylabel('Probability')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#LAB PROGRAM 2 (Question 7)
p_geo = 0.05
geo = stats.geom(p_geo)
geo.mean()
geo.pmf(1)
1 - geo.cdf(5)
print(f'Geometric(p={p_geo}):')
print(f'  Mean = 1/p = {geo.mean():.4f}')
print(f'  P(X=1) = {geo.pmf(1):.4f}')
print(f'  P(X>5) = {1 - geo.cdf(5):.4f}')

# Visualization

plt.figure(figsize=(7, 5))
x_g = np.arange(1, 20)
plt.bar(x_g, geo.pmf(x_g), color='green',edgecolor='black', alpha=0.8)
plt.title(f'Geometric PMF (p={p_geo})')
plt.xlabel('Trial Number (x)')
plt.ylabel('Probability')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
